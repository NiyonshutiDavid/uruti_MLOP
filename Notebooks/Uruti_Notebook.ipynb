To reliably obtain a loadable audio file, I will generate a simple synthetic WAV file locally using `numpy` and `scipy.io.wavfile`, ensuring it's a valid and accessible audio source for `librosa`.



```python
import librosa
import librosa.display
import numpy as np
from scipy.io.wavfile import write as write_wav

# Generate a simple sine wave audio
sr_gen = 22050 # Sample rate
duration = 3   # seconds
f_freq = 440   # Hz (A4 note)
t = np.linspace(0., duration, int(sr_gen * duration), endpoint=False)
y_gen = 0.5 * np.sin(2 * np.pi * f_freq * t)

# Normalize to 16-bit integer values for WAV export
y_gen_int = np.int16(y_gen * 32767)

# Save the generated audio to a WAV file
audio_path = 'generated_sample.wav'
write_wav(audio_path, sr_gen, y_gen_int)

print(f"Generated audio saved to: {audio_path}")

# Load the generated audio file using librosa
y, sr = librosa.load(audio_path)

print(f"Audio loaded successfully: {audio_path}")
print(f"Audio data shape: {y.shape}")
print(f"Sampling rate (sr): {sr} Hz")
```

Now that the audio waveform is visualized, I will proceed to calculate the spectrogram using `librosa.stft` and convert it to decibels using `librosa.amplitude_to_db` as specified in the instructions. This is a crucial preprocessing step for analyzing frequency content over time.



```python
import librosa
import librosa.display
import matplotlib.pyplot as plt

# Calculate the Short-Time Fourier Transform (STFT) to get the spectrogram
D = librosa.stft(y)

# Convert the amplitude spectrogram to decibels for visualization
S_db = librosa.amplitude_to_db(np.abs(D), ref=np.max)

# Visualize the spectrogram
plt.figure(figsize=(14, 5))
librosa.display.specshow(S_db, sr=sr, x_axis='time', y_axis='log')
plt.colorbar(format='%+2.0f dB')
plt.title('Spectrogram (dB)')
plt.xlabel('Time (s)')
plt.ylabel('Frequency (Hz)')
plt.show()
```

```python
import librosa.feature
import librosa.display
import matplotlib.pyplot as plt

# Extract MFCCs from the audio signal
mfccs = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13) # n_mfcc is typically 13

# Visualize the MFCCs
plt.figure(figsize=(14, 5))
librosa.display.specshow(mfccs, sr=sr, x_axis='time')
plt.colorbar()
plt.title('MFCCs')
plt.xlabel('Time (s)')
plt.ylabel('MFCC Coefficients')
plt.tight_layout()
plt.show()
```

Combine the extracted audio features (MFCCs, ZCR, Spectral Centroid, Spectral Bandwidth) into a unified structure, and calculate descriptive statistics for these features across the entire audio signal.


**Reasoning**:
To prepare the MFCCs for combination with other features in a DataFrame, I need to transpose them so that each row represents a time frame, aligning with the ZCR, Spectral Centroid, and Spectral Bandwidth dimensions.



```python
import pandas as pd
import numpy as np

# Transpose MFCCs so that each row is a frame and each column is an MFCC coefficient
mfccs_transposed = mfccs.T

# Ensure all features have the same number of frames
min_frames = min(mfccs_transposed.shape[0], len(zcr), len(spectral_centroids), len(spectral_bandwidth))

# Trim features to the minimum number of frames to ensure alignment
mfccs_trimmed = mfccs_transposed[:min_frames, :]
zcr_trimmed = zcr[:min_frames]
spectral_centroids_trimmed = spectral_centroids[:min_frames]
spectral_bandwidth_trimmed = spectral_bandwidth[:min_frames]

# Create a dictionary to store features for the DataFrame
feature_dict = {}
for i in range(mfccs_trimmed.shape[1]):
    feature_dict[f'MFCC_{i+1}'] = mfccs_trimmed[:, i]

feature_dict['ZCR'] = zcr_trimmed
feature_dict['Spectral_Centroid'] = spectral_centroids_trimmed
feature_dict['Spectral_Bandwidth'] = spectral_bandwidth_trimmed

# Create the DataFrame
audio_features_df = pd.DataFrame(feature_dict)

print("Audio features DataFrame created with shape:", audio_features_df.shape)
print("First 5 rows of the DataFrame:")
print(audio_features_df.head())
```

## Video Data Processing and Visualization

### Subtask:
Develop code to load video datasets, extract key frames, and process them for feature extraction (e.g., using pre-trained CNNs). Visualize key frames and their extracted features.


**Reasoning**:
To begin the video data processing subtask, I will first download a sample video file using `!wget`, then load it using `cv2.VideoCapture`. I will extract and print its properties, and finally extract and visualize a few key frames after converting them to RGB format for proper display.



```python
import cv2
import matplotlib.pyplot as plt

# 1. Download a sample video file
!wget -nc https://www.learningcontainer.com/wp-content/uploads/2020/05/sample-mp4-file.mp4 -O sample_video.mp4

# 2. Load the downloaded video
video_path = 'sample_video.mp4'
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print(f"Error: Could not open video file {video_path}")
else:
    print(f"Successfully opened video file: {video_path}")

    # 3. Get basic video properties
    frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    fps = cap.get(cv2.CAP_PROP_FPS)
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

    print(f"Total frames: {frame_count}")
    print(f"Frames per second (FPS): {fps}")
    print(f"Frame width: {width}")
    print(f"Frame height: {height}")

    # 4. Extract and store key frames
    extracted_frames = []
    frame_interval = max(1, frame_count // 10) # Extract up to 10 frames or fewer if video is short

    for i in range(frame_count):
        ret, frame = cap.read()
        if not ret:
            break
        if i % frame_interval == 0 and len(extracted_frames) < 10: # Limit to max 10 frames
            # Convert BGR to RGB for matplotlib display
            frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            extracted_frames.append(frame_rgb)

    print(f"Extracted {len(extracted_frames)} key frames.")

    # 5. Visualize the extracted key frames
    plt.figure(figsize=(15, 5))
    for i, frame in enumerate(extracted_frames):
        plt.subplot(1, len(extracted_frames), i + 1)
        plt.imshow(frame)
        plt.title(f"Frame {i * frame_interval}")
        plt.axis('off')
    plt.tight_layout()
    plt.show()

    cap.release()
    print("Video capture released.")
```

**Reasoning**:
Now that key frames have been extracted and visualized, the next step is to process these frames for feature extraction using a pre-trained Convolutional Neural Network (CNN). I will use MobileNetV2 from `tensorflow.keras.applications` to extract high-level features from each key frame. First, the frames will be preprocessed to match the input requirements of the CNN, then features will be extracted, and finally, a sample of these features will be visualized as a heatmap to understand their structure.



```python
import tensorflow as tf
from tensorflow.keras.applications.mobilenet_v2 import MobileNetV2, preprocess_input
import numpy as np
import matplotlib.pyplot as plt

# 1. Load a pre-trained CNN model (MobileNetV2 without the top classification layer)
# This will allow us to use it as a feature extractor
base_model = MobileNetV2(weights='imagenet', include_top=False, input_shape=(224, 224, 3))
model = tf.keras.Model(inputs=base_model.input, outputs=base_model.output)
print("Pre-trained MobileNetV2 model loaded.")

# 2. Preprocess and extract features from the extracted_frames
features_list = []
for i, frame_rgb in enumerate(extracted_frames):
    # Resize frame to the input size expected by MobileNetV2 (224x224)
    resized_frame = cv2.resize(frame_rgb, (224, 224))
    # Expand dimensions to create a batch (1, 224, 224, 3)
    input_frame = np.expand_dims(resized_frame, axis=0)
    # Preprocess the input for MobileNetV2
    preprocessed_input = preprocess_input(input_frame)
    # Extract features
    features = model.predict(preprocessed_input)
    # Flatten the features for easier storage and potential visualization later
    features_list.append(features.flatten())

print(f"Extracted features for {len(features_list)} frames.")
print(f"Example feature vector shape: {features_list[0].shape}")

# 3. Visualize a sample of the extracted features (e.g., for the first frame) as a heatmap
if features_list:
    sample_features = features_list[0]
    plt.figure(figsize=(15, 3))
    plt.imshow(sample_features.reshape(1, -1), cmap='viridis', aspect='auto')
    plt.colorbar()
    plt.title('Feature Vector Heatmap for First Key Frame')
    plt.xlabel('Feature Dimension')
    plt.ylabel('Frame')
    plt.yticks([]) # Hide y-axis ticks as there's only one row
    plt.tight_layout()
    plt.show()

print("Feature extraction and visualization complete for key frames.")
```

```python
training_data = [
    {"prompt": "Hello, how are you?", "response": "I am doing well, thank you! How can I assist you today?"},
    {"prompt": "What is the capital of France?", "response": "The capital of France is Paris."},
    {"prompt": "Tell me about the URUTI platform.", "response": "The URUTI platform is designed to develop advanced AI models, including a Llama3 chatbot, an MLP for startup analysis, and a Reinforcement Learning-based pitch coach."},
    {"prompt": "Can you help me write an email?", "response": "Yes, I can. What is the topic of the email?"},
    {"prompt": "Who developed Llama3?", "response": "Llama3 is developed by Meta."},
    {"prompt": "What is your purpose?", "response": "I am an AI assistant designed to help with a wide range of tasks and provide information."}
]

print(f"Prepared a synthetic dataset with {len(training_data)} conversation turns.")
print("Example conversation turn:")
print(training_data[0])
```

**Reasoning**:
Now that the synthetic dataset is prepared, I will tokenize it using the previously loaded tokenizer.
```python
import torch

# Tokenize the prepared dataset
tokenized_datasets = []
max_length = 128 # Define a suitable max length for the tokenizer

for item in training_data:
    # Concatenate prompt and response for training a causal language model
    full_text = f"### Instruction:\n{item['prompt']}\n\n### Response:\n{item['response']}"

    # Tokenize the full text
    # The `truncation=True` ensures that sequences longer than max_length are truncated.
    # The `padding='max_length'` ensures all sequences have the same length.
    inputs = tokenizer(full_text, return_tensors='pt', truncation=True, padding='max_length', max_length=max_length)
    tokenized_datasets.append(inputs)

print(f"Tokenized {len(tokenized_datasets)} conversation turns.")
print("Example tokenized input (first item):")
print(f"  Input IDs shape: {tokenized_datasets[0]['input_ids'].shape}")
print(f"  Attention Mask shape: {tokenized_datasets[0]['attention_mask'].shape}")
print(f"  Decoded text (first 50 tokens): {tokenizer.decode(tokenized_datasets[0]['input_ids'][0][:50])}")
```

```python
import numpy as np
import pandas as pd

# 1. Define the number of samples
n_samples = 1000

# 2. Generate synthetic data for features
np.random.seed(42) # for reproducibility

funding_rounds = np.random.randint(1, 6, n_samples)  # 1 to 5 rounds
team_size = np.random.randint(2, 51, n_samples)      # 2 to 50 employees
market_size = np.random.rand(n_samples) * 100        # 0 to 100 billion
innovation_score = np.random.rand(n_samples) * 10    # 0 to 10
competition_level = np.random.rand(n_samples)        # 0 to 1 (high competition to low)

# 3. Generate a target variable: success_score (continuous for regression)
# Influenced by features with some added noise
success_score = (
    funding_rounds * 0.5
    + team_size * 0.1
    + market_size * 0.3
    + innovation_score * 0.8
    - competition_level * 10
    + np.random.randn(n_samples) * 5 # Add Gaussian noise
)

# Normalize success_score to a more interpretable range, e.g., 0-100
success_score = (success_score - success_score.min()) / (success_score.max() - success_score.min()) * 100

# 4. Combine into a Pandas DataFrame
startup_df = pd.DataFrame({
    'funding_rounds': funding_rounds,
    'team_size': team_size,
    'market_size': market_size,
    'innovation_score': innovation_score,
    'competition_level': competition_level,
    'success_score': success_score
})

print("Generated synthetic startup dataset.")
print("DataFrame head:")
print(startup_df.head())

print("\nDataFrame info:")
startup_df.info()
```

This class will simulate the interaction between a presenter and a coach, managing state transitions and reward calculations based on actions, as specified in the subtask instructions.



```python
import numpy as np

class PitchEnvironment:
    def __init__(self):
        # Initial state (hypothetical features for a presenter's pitch)
        # Ranges: Pace_Score (0-1), Tone_Score (0-1), Sentiment_Score (-1 to 1), Body_Language_Score (0-1), Content_Clarity_Score (0-1)
        self.state = {
            'Pace_Score': np.random.uniform(0.3, 0.7), # Start near optimal for demonstration
            'Tone_Score': np.random.uniform(0.3, 0.7),
            'Sentiment_Score': np.random.uniform(-0.2, 0.2),
            'Body_Language_Score': np.random.uniform(0.4, 0.6),
            'Content_Clarity_Score': np.random.uniform(0.4, 0.6)
        }
        self.optimal_ranges = {
            'Pace_Score': (0.4, 0.6),
            'Tone_Score': (0.4, 0.6),
            'Sentiment_Score': (-0.1, 0.1), # Neutral is optimal
            'Body_Language_Score': (0.7, 0.9),
            'Content_Clarity_Score': (0.7, 0.9)
        }
        self.current_step = 0
        self.max_steps = 10 # Simulate a pitch of 10 interaction steps

        # Action definitions (mapping from action index to effect)
        self.action_effects = {
            0: 'No_Action',
            1: 'Adjust_Pace_Slower',
            2: 'Adjust_Pace_Faster',
            3: 'Be_More_Energetic',
            4: 'Be_Calmer',
            5: 'Improve_Sentiment_Positive',
            6: 'Adjust_Body_Language_Improve',
            7: 'Clarify_Content'
        }
        self.action_space_n = len(self.action_effects)

    def _is_optimal(self, feature_name, value):
        low, high = self.optimal_ranges[feature_name]
        return low <= value <= high

    def _get_overall_optimality_score(self):
        # A simple score for how close the pitch is to optimal conditions
        score = 0
        for feature, val in self.state.items():
            low, high = self.optimal_ranges[feature]
            if self._is_optimal(feature, val):
                score += 1
            else:
                # Penalize based on deviation from optimal range midpoint
                midpoint = (low + high) / 2
                deviation = abs(val - midpoint)
                score -= deviation * 2 # Heuristic penalty
        return score

    def step(self, action):
        reward = 0
        done = False

        # Store previous state for reward calculation
        prev_state = self.state.copy()

        # Apply action effects (simplified dynamics)
        if action == 0: # No_Action
            # Small penalty if not optimal, small reward if already optimal
            if self._get_overall_optimality_score() < len(self.state) * 0.7: # If not mostly optimal
                reward -= 1
            else:
                reward += 2
        elif action == 1: # Adjust_Pace_Slower
            if self.state['Pace_Score'] > self.optimal_ranges['Pace_Score'][1]: # Too fast, improving
                self.state['Pace_Score'] = max(0, self.state['Pace_Score'] - np.random.uniform(0.05, 0.15))
                reward += 15
            elif self.state['Pace_Score'] <= self.optimal_ranges['Pace_Score'][0]: # Already too slow or optimal, worsening
                self.state['Pace_Score'] = max(0, self.state['Pace_Score'] - np.random.uniform(0.05, 0.15))
                reward -= 10
            else: # Optimal, but still acting, small penalty
                self.state['Pace_Score'] = max(0, self.state['Pace_Score'] - np.random.uniform(0.01, 0.05))
                reward -= 5
        elif action == 2: # Adjust_Pace_Faster
            if self.state['Pace_Score'] < self.optimal_ranges['Pace_Score'][0]: # Too slow, improving
                self.state['Pace_Score'] = min(1, self.state['Pace_Score'] + np.random.uniform(0.05, 0.15))
                reward += 15
            elif self.state['Pace_Score'] >= self.optimal_ranges['Pace_Score'][1]: # Already too fast or optimal, worsening
                self.state['Pace_Score'] = min(1, self.state['Pace_Score'] + np.random.uniform(0.05, 0.15))
                reward -= 10
            else: # Optimal, but still acting, small penalty
                self.state['Pace_Score'] = min(1, self.state['Pace_Score'] + np.random.uniform(0.01, 0.05))
                reward -= 5
        elif action == 3: # Be_More_Energetic
            if self.state['Tone_Score'] < self.optimal_ranges['Tone_Score'][0]: # Too low, improving
                self.state['Tone_Score'] = min(1, self.state['Tone_Score'] + np.random.uniform(0.05, 0.15))
                reward += 20
            elif self.state['Tone_Score'] >= self.optimal_ranges['Tone_Score'][1]: # Too energetic, worsening
                self.state['Tone_Score'] = min(1, self.state['Tone_Score'] + np.random.uniform(0.05, 0.15))
                reward -= 15
            else:
                self.state['Tone_Score'] = min(1, self.state['Tone_Score'] + np.random.uniform(0.01, 0.05))
                reward -= 5
        elif action == 4: # Be_Calmer
            if self.state['Tone_Score'] > self.optimal_ranges['Tone_Score'][1]: # Too high, improving
                self.state['Tone_Score'] = max(0, self.state['Tone_Score'] - np.random.uniform(0.05, 0.15))
                reward += 10
            elif self.state['Tone_Score'] <= self.optimal_ranges['Tone_Score'][0]: # Too calm, worsening
                self.state['Tone_Score'] = max(0, self.state['Tone_Score'] - np.random.uniform(0.05, 0.15))
                reward -= 10
            else:
                self.state['Tone_Score'] = max(0, self.state['Tone_Score'] - np.random.uniform(0.01, 0.05))
                reward -= 5
        elif action == 5: # Improve_Sentiment_Positive
            if self.state['Sentiment_Score'] < self.optimal_ranges['Sentiment_Score'][0]: # Too negative, improving
                self.state['Sentiment_Score'] = min(1, self.state['Sentiment_Score'] + np.random.uniform(0.1, 0.3))
                reward += 15
            else: # Already neutral or positive, neutral effect or slight penalty for unnecessary action
                self.state['Sentiment_Score'] = min(1, self.state['Sentiment_Score'] + np.random.uniform(0.01, 0.05))
                reward -= 5
        elif action == 6: # Adjust_Body_Language_Improve
            if self.state['Body_Language_Score'] < self.optimal_ranges['Body_Language_Score'][0]: # Needs improvement
                self.state['Body_Language_Score'] = min(1, self.state['Body_Language_Score'] + np.random.uniform(0.05, 0.15))
                reward += 20
            else: # Already good, small penalty
                self.state['Body_Language_Score'] = min(1, self.state['Body_Language_Score'] + np.random.uniform(0.01, 0.05))
                reward -= 5
        elif action == 7: # Clarify_Content
            if self.state['Content_Clarity_Score'] < self.optimal_ranges['Content_Clarity_Score'][0]: # Needs clarity
                self.state['Content_Clarity_Score'] = min(1, self.state['Content_Clarity_Score'] + np.random.uniform(0.05, 0.15))
                reward += 20
            else: # Already clear, small penalty
                self.state['Content_Clarity_Score'] = min(1, self.state['Content_Clarity_Score'] + np.random.uniform(0.01, 0.05))
                reward -= 5

        # Clamp state values to their valid ranges
        self.state['Pace_Score'] = np.clip(self.state['Pace_Score'], 0.0, 1.0)
        self.state['Tone_Score'] = np.clip(self.state['Tone_Score'], 0.0, 1.0)
        self.state['Sentiment_Score'] = np.clip(self.state['Sentiment_Score'], -1.0, 1.0)
        self.state['Body_Language_Score'] = np.clip(self.state['Body_Language_Score'], 0.0, 1.0)
        self.state['Content_Clarity_Score'] = np.clip(self.state['Content_Clarity_Score'], 0.0, 1.0)


        self.current_step += 1

        # Check if episode is done
        if self.current_step >= self.max_steps or self._get_overall_optimality_score() >= len(self.state) * 0.9: # End if optimal enough
            done = True
            if self._get_overall_optimality_score() >= len(self.state) * 0.9:
                reward += 50 # Large positive terminal reward
            else:
                reward -= 20 # Negative terminal reward if max steps reached without optimality

        # Convert state dictionary to a numpy array for RL agent
        obs = np.array(list(self.state.values()), dtype=np.float32)

        return obs, reward, done, {}

    def reset(self):
        self.state = {
            'Pace_Score': np.random.uniform(0.3, 0.7),
            'Tone_Score': np.random.uniform(0.3, 0.7),
            'Sentiment_Score': np.random.uniform(-0.2, 0.2),
            'Body_Language_Score': np.random.uniform(0.4, 0.6),
            'Content_Clarity_Score': np.random.uniform(0.4, 0.6)
        }
        self.current_step = 0
        return np.array(list(self.state.values()), dtype=np.float32)

# Example usage:
env = PitchEnvironment()
initial_obs = env.reset()
print("Initial State:", env.state)
print("Initial Observation (array):", initial_obs)

obs, reward, done, _ = env.step(action=1) # Agent suggests to slow pace
print("\nAfter 1 step (action=1):")
print("New State:", env.state)
print("Reward:", reward)
print("Done:", done)
print("Observation (array):", obs)
```

```python
from stable_baselines3.common.env_util import make_vec_env
from stable_baselines3.common.callbacks import BaseCallback
import matplotlib.pyplot as plt
import numpy as np

# Custom Callback to log episode rewards and lengths
class CustomCallback(BaseCallback):
    def __init__(self, verbose=0):
        super(CustomCallback, self).__init__(verbose)
        self.ep_rewards = []
        self.ep_lengths = []
        self.current_reward_sum = 0
        self.current_ep_len = 0

    def _on_step(self) -> bool:
        # Check if the episode is done
        if self.locals['dones'][0]:
            # Record the reward and length for the finished episode
            self.ep_rewards.append(self.current_reward_sum + self.locals['rewards'][0])
            self.ep_lengths.append(self.current_ep_len + 1)
            # Reset for the next episode
            self.current_reward_sum = 0
            self.current_ep_len = 0
        else:
            # Accumulate reward and length during the episode
            self.current_reward_sum += self.locals['rewards'][0]
            self.current_ep_len += 1
        return True

# It's often good practice to wrap the environment in a VecEnv for stable-baselines3
# For a single environment, DummyVecEnv is sufficient
vec_env = make_vec_env(CustomPitchEnv, n_envs=1)

# Instantiate the PPO agent
model = PPO("MlpPolicy", vec_env, verbose=0) # Set verbose to 0 to avoid clutter during callback logging

print("PPO agent instantiated.")

# Create the custom callback instance
callback = CustomCallback()

# Train the agent with the callback
print("\nStarting PPO agent training with callback...")
model.learn(total_timesteps=5000, callback=callback)
print("PPO agent training complete.")

# Visualize the learning curve (episode rewards)
plt.figure(figsize=(12, 6))
plt.plot(callback.ep_rewards)
plt.xlabel('Episode')
plt.ylabel('Cumulative Reward')
plt.title('PPO Agent Learning Curve (Cumulative Reward per Episode)')
plt.grid(True)
plt.show()

# Visualize the episode lengths (optional)
plt.figure(figsize=(12, 6))
plt.plot(callback.ep_lengths)
plt.xlabel('Episode')
plt.ylabel('Episode Length')
plt.title('PPO Agent Episode Lengths')
plt.grid(True)
plt.show()
```
